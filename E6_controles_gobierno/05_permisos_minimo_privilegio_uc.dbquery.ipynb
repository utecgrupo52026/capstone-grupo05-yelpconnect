```sql
-- ============================================================
-- YELPCONNECT
-- 05 · PERMISOS DE MÍNIMO PRIVILEGIO — UNITY CATALOG
-- ============================================================
--
-- Objetivo:
-- Aplicar acceso mínimo al grupo analítico de prueba.
--
-- Principio:
-- El usuario consume únicamente productos Gold certificados.
-- No recibe acceso directo a Bronze, Silver, Gov ni productos
-- sensibles de Trust & Safety.
-- ============================================================


-- ------------------------------------------------------------
-- 1. ACCESO AL CATÁLOGO
-- ------------------------------------------------------------

GRANT USE CATALOG
ON CATALOG workspace
TO `grp_analytics_test`;


-- ------------------------------------------------------------
-- 2. ACCESO AL SCHEMA GOLD
-- ------------------------------------------------------------

GRANT USE SCHEMA
ON SCHEMA workspace.yelp_gold
TO `grp_analytics_test`;


-- ------------------------------------------------------------
-- 3. ACCESO ÚNICAMENTE AL PRODUCTO ANALÍTICO CERTIFICADO
-- ------------------------------------------------------------

GRANT SELECT
ON TABLE workspace.yelp_gold.gld_kpi_trust_safety_overview
TO `grp_analytics_test`;


-- ============================================================
-- 4. EVIDENCIA DE PERMISOS
-- ============================================================

SHOW GRANTS ON CATALOG workspace;

SHOW GRANTS ON SCHEMA workspace.yelp_gold;

SHOW GRANTS ON TABLE
workspace.yelp_gold.gld_kpi_trust_safety_overview;

SHOW GRANTS ON TABLE
workspace.yelp_gold.gld_investigation_queue;
```
