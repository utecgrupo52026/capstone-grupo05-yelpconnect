```sql
-- ============================================================
-- YELPCONNECT
-- EVIDENCIA PRE-RUN · JOB OPERATIVO
-- ============================================================


-- ============================================================
-- 1. ÚLTIMOS EVALUATION RUNS
-- ============================================================

SELECT
    evaluation_run_id,
    started_at,
    finished_at,
    execution_status,
    dataset_version,
    pipeline_name
FROM workspace.yelp_gov.evaluation_run
ORDER BY started_at DESC
LIMIT 5;


-- ============================================================
-- 2. BRONZE
-- ============================================================

SELECT
    'brz_review' AS objeto,
    COUNT(*) AS filas
FROM workspace.yelp_bronze.brz_review;


-- ============================================================
-- 3. SILVER BASE
-- ============================================================

SELECT
    'slv_review' AS objeto,
    COUNT(*) AS filas
FROM workspace.yelp_silver.slv_review;


SELECT
    'slv_review_features' AS objeto,
    COUNT(*) AS filas
FROM workspace.yelp_silver.slv_review_features;


-- ============================================================
-- 4. ÚLTIMO RUN T&S EXITOSO
-- ============================================================

WITH ultimo_run AS (

    SELECT
        evaluation_run_id

    FROM workspace.yelp_gov.evaluation_run

    WHERE execution_status = 'SUCCESS'

    ORDER BY started_at DESC

    LIMIT 1
)

SELECT
    evaluation_run_id,
    COUNT(*) AS evaluaciones

FROM workspace.yelp_silver.slv_review_evaluation

WHERE evaluation_run_id = (
    SELECT evaluation_run_id
    FROM ultimo_run
)

GROUP BY evaluation_run_id;


-- ============================================================
-- 5. RISK DEL ÚLTIMO RUN EXITOSO
-- ============================================================

WITH ultimo_run AS (

    SELECT
        evaluation_run_id

    FROM workspace.yelp_gov.evaluation_run

    WHERE execution_status = 'SUCCESS'

    ORDER BY started_at DESC

    LIMIT 1
)

SELECT
    evaluation_run_id,
    COUNT(*) AS reviews_risk

FROM workspace.yelp_silver.slv_review_risk

WHERE evaluation_run_id = (
    SELECT evaluation_run_id
    FROM ultimo_run
)

GROUP BY evaluation_run_id;


-- ============================================================
-- 6. INVESTIGATION QUEUE
-- ============================================================

SELECT
    COUNT(*) AS investigation_queue

FROM workspace.yelp_gold.gld_investigation_queue;


-- ============================================================
-- 7. KPI EJECUTIVO GOLD
-- ============================================================

SELECT *
FROM workspace.yelp_gold.gld_kpi_trust_safety_overview;


-- ============================================================
-- 8. ÚLTIMA EVIDENCIA DQ
-- ============================================================

SELECT *
FROM workspace.yelp_gov.dq_resultados
ORDER BY ejecucion_ts DESC
LIMIT 50;
```
